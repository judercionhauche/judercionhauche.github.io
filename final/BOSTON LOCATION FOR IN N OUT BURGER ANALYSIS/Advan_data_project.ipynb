```python
import json
import ast
import pandas as pd

def parse_bucketed_dwell(x):
    if pd.isna(x):
        return None

    # 1st attempt: JSON load once
    try:
        obj = json.loads(x)
    except Exception:
        obj = x

    # If we got a dict already → done
    if isinstance(obj, dict):
        return obj

    # If we got a string that still looks like JSON → try loading again
    if isinstance(obj, str):
        s = obj.strip()
        # Try JSON again
        try:
            obj2 = json.loads(s)
            if isinstance(obj2, dict):
                return obj2
        except Exception:
            pass
        # Try Python-style dict (single quotes, etc.)
        try:
            obj2 = ast.literal_eval(s)
            if isinstance(obj2, dict):
                return obj2
        except Exception:
            pass

    return None
```

```python
import json

def parse_pop_hours(x):
    # Already a list? keep it
    if isinstance(x, list):
        return x
    
    if not isinstance(x, str):
        return None
    
    s = x
    
    # Try up to 2 JSON decodes to handle the double-encoded case
    for _ in range(2):
        try:
            s = json.loads(s)
        except json.JSONDecodeError:
            break
    
    # If after decoding we have a list, return it
    if isinstance(s, list):
        return s
    
    return None

df_rest["POP_HOUR_LIST"] = df_rest["POPULARITY_BY_HOUR"].apply(parse_pop_hours)
```

```python
import numpy as np

def get_peak_hour(lst):
    if isinstance(lst, list) and len(lst) == 24:
        return int(np.argmax(lst))
    return np.nan

def evening_share(lst):
    if not (isinstance(lst, list) and len(lst) == 24):
        return np.nan
    total = sum(lst)
    if total == 0:
        return 0
    # hours 17–22 (17,18,19,20,21,22)
    return sum(lst[17:23]) / total

df_rest["peak_hour"] = df_rest["POP_HOUR_LIST"].apply(get_peak_hour)
df_rest["evening_share"] = df_rest["POP_HOUR_LIST"].apply(evening_share)
```

```python
def parse_visits_by_day(x):
    if pd.isna(x):
        return None

    # First attempt: JSON once
    try:
        obj = json.loads(x)
    except Exception:
        obj = x

    # If we already have a list → good
    if isinstance(obj, list):
        return obj

    # If we got a dict (rare, but just in case) → use its values
    if isinstance(obj, dict):
        return list(obj.values())

    # If it's still a string, try again
    if isinstance(obj, str):
        s = obj.strip()
        # Try JSON again
        try:
            obj2 = json.loads(s)
            if isinstance(obj2, list):
                return obj2
            if isinstance(obj2, dict):
                return list(obj2.values())
        except Exception:
            pass
        # Try Python-style literal (single quotes, etc.)
        try:
            obj2 = ast.literal_eval(s)
            if isinstance(obj2, list):
                return obj2
            if isinstance(obj2, dict):
                return list(obj2.values())
        except Exception:
            pass

    return None
```

```python
df_rest["VISITS_BY_DAY_LIST"] = df_rest["VISITS_BY_DAY"].apply(parse_visits_by_day)
```

```python
def safe_avg(lst):
    if isinstance(lst, list) and len(lst) > 0:
        return float(np.mean(lst))
    return np.nan

def safe_max(lst):
    if isinstance(lst, list) and len(lst) > 0:
        return float(np.max(lst))
    return np.nan

def safe_std(lst):
    if isinstance(lst, list) and len(lst) > 1:
        return float(np.std(lst))
    return np.nan

df_rest["avg_daily_visits"] = df_rest["VISITS_BY_DAY_LIST"].apply(safe_avg)
df_rest["max_daily_visits"] = df_rest["VISITS_BY_DAY_LIST"].apply(safe_max)
df_rest["std_daily_visits"] = df_rest["VISITS_BY_DAY_LIST"].apply(safe_std)
```

```python
def minmax(s):
    s = s.astype(float)
    min_v, max_v = s.min(), s.max()
    if pd.isna(min_v) or pd.isna(max_v) or max_v == min_v:
        return pd.Series(0.5, index=s.index)  # neutral if constant
    return (s - min_v) / (max_v - min_v)
```

```python
df_rest["visit_volume_score"] = minmax(df_rest["RAW_VISIT_COUNTS"].fillna(0))
```

```python
# First normalise std
std_norm = minmax(df_rest["std_daily_visits"].fillna(0))

# Stability = 1 - normalized volatility
df_rest["stability_score"] = 1 - std_norm
```

```python
density_norm = minmax(df_rest["zip_restaurant_count"])
df_rest["competition_score"] = 1 - density_norm
```

```python
score_cols_to_norm = [
    "fast_food_behavior_score",
    "weekend_demand_score",
    "weekday_daytime_score",
    "visit_volume_score",
    "stability_score",
    "competition_score"
]

for col in score_cols_to_norm:
    df_rest[col + "_norm"] = minmax(df_rest[col].fillna(0))
```
